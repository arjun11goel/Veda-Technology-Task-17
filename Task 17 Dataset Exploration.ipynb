# Dataset Exploration - Task 17

**Objective:** Understand the structure and basic characteristics of a dataset before modeling.

We will use the **Iris Dataset** for this exploration.

## Step 1: Import Libraries
First, we need to import the Pandas library, which is essential for data manipulation and analysis in Python.

```python
import pandas as pd
```

## Step 2: Load the Dataset
We load the Iris dataset into a Pandas DataFrame. Make sure the `iris_dataset.csv` file is in the same directory as this notebook.

```python
df = pd.read_csv('iris_dataset.csv')
```

## Step 3: View Dataset Head and Tail
Let's look at the first and last few rows to get a quick sense of what the data looks like.

```python
print("--- First 5 rows ---")
display(df.head())

print("\n--- Last 5 rows ---")
display(df.tail())
```

## Step 4: Dataset Dimensions
The `.shape` attribute returns a tuple representing the dimensionality of the DataFrame (rows, columns).

```python
print(f"The dataset has {df.shape[0]} rows and {df.shape[1]} columns.")
```

## Step 5: Column Information
Let's list all the column names to understand the features present in the dataset.

```python
print("Columns in the dataset:")
print(df.columns.tolist())
```

## Step 6: Identify Data Types and Non-Null Counts
The `.info()` method prints a concise summary of a DataFrame, including the index dtype, column dtypes, non-null values, and memory usage. This helps identify numerical and categorical columns.

```python
df.info()
```

## Step 7: Examine Numerical Statistics
The `.describe()` method generates descriptive statistics for the numerical columns, summarizing the central tendency, dispersion, and shape of the dataset's distribution.

```python
df.describe()
```

## Step 8: Five Observations About the Dataset
Based on the exploration above, here are five observations:
1.  **Size:** The dataset consists of 150 rows (samples) and 5 columns (features).
2.  **Features:** The four numerical features are sepal length, sepal width, petal length, and petal width, all measured in centimeters.
3.  **Target Variable:** There is one categorical column, typically named 'species' or 'class', representing the type of Iris flower.
4.  **Completeness:** There appear to be no missing values (nulls) in the dataset, as all columns have 150 non-null counts.
5.  **Scale:** Petal lengths have a wider range (1.0 to 6.9 cm) and higher standard deviation compared to sepal widths, which are more tightly clustered (2.0 to 4.4 cm).

## Interview Questions Answers

**1. Why should a dataset be explored before modeling?**
Exploring a dataset (Exploratory Data Analysis or EDA) is crucial before modeling because it helps you understand the data's underlying structure, detect anomalies or outliers, identify missing values, and uncover relationships between variables. This understanding dictates the data preprocessing steps needed and helps in selecting the appropriate machine learning algorithms, ultimately leading to more accurate and reliable models.

**2. What does DataFrame.shape return?**
`DataFrame.shape` returns a tuple representing the dimensionality of the DataFrame. The format is `(number_of_rows, number_of_columns)`.

**3. What information does info() provide?**
The `info()` method provides a concise summary of the DataFrame. It prints information about the index and column data types, the count of non-null values in each column, and the total memory usage of the DataFrame.
